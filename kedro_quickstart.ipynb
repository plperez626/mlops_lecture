# Kedro Quickstart

A quick hands-on introduction to Kedro - build and run a simple data pipeline in 5 minutes.

## Step 1: Install Kedro & Create Project

````python
# Create project
!kedro new --name=my_kedro_project --tools=none --example=n

import os
print("✓ Kedro installed and project created!" if os.path.exists('my-kedro-project') else "⚠ Setup failed")
````

## Step 2: Create a Pipeline

Define a simple data processing function and configure it as a Kedro pipeline.

````python
import os

# Create pipeline directory and files
pipeline_dir = 'my-kedro-project/src/my_kedro_project/pipelines/data_processing'
os.makedirs(pipeline_dir, exist_ok=True)

# Define a simple data processing function
with open(f'{pipeline_dir}/nodes.py', 'w') as f:
    f.write('''def process_data(value: float, multiplier: float) -> float:
    """Simple data processing function."""
    result = value * multiplier
    print(f"Processing: {value} × {multiplier} = {result}")
    return result
''')

# Create the pipeline
with open(f'{pipeline_dir}/pipeline.py', 'w') as f:
    f.write('''from kedro.pipeline import Pipeline, node
from .nodes import process_data

def create_pipeline(**kwargs) -> Pipeline:
    return Pipeline([
        node(func=process_data,
             inputs=["params:input_value", "params:multiplier"],
             outputs="result",
             name="process_data_node"),
    ])
''')

with open(f'{pipeline_dir}/__init__.py', 'w') as f:
    f.write('')

# Set parameters
os.makedirs('my-kedro-project/conf/base', exist_ok=True)
with open('my-kedro-project/conf/base/parameters.yml', 'w') as f:
    f.write('input_value: 42\nmultiplier: 2.5\n')

print("✓ Pipeline created with a simple data processing node")
````

## Step 3: Run the Pipeline

````python
import os

if not os.path.exists('my-kedro-project'):
    print("⚠ Run Step 1 first to create the project!")
else:
    os.chdir('my-kedro-project')
    
    # Register the pipeline
    os.makedirs('src/my_kedro_project', exist_ok=True)
    with open('src/my_kedro_project/pipeline_registry.py', 'w') as f:
        f.write('''from kedro.pipeline import Pipeline
from my_kedro_project.pipelines.data_processing.pipeline import create_pipeline

def register_pipelines():
    return {"__default__": create_pipeline()}
''')
    
    print("Running pipeline...\n")
    !kedro run
    
    os.chdir('..')
````

## Step 4 : Visualize the Kedro pipeline

1. Open a terminal and go to the project folder:

```bash
cd my-kedro-project
```

2. Start Kedro-Viz with uv:

```bash
uv run kedro viz run
```

3. Kedro-Viz starts a local server and usually opens your browser automatically. If it does not, open this address manually:

```
http://127.0.0.1:4141
```

4. In the page, you can observe the pipeline: nodes (Python functions), datasets, and the links between them.   
    You can zoom, click a node to see its details, and switch between pipelines from the menu.

5. To stop the server, go back to the terminal and press Ctrl+C.

You can precise the port with the `--port` parameter.
```bash
uv run kedro viz run --port 4242
```
