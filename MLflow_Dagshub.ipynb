# From dataset to production: an end-to-end MLOps demo

Welcome! This notebook is a demonstration of the full lifecycle of a machine learning model: we start by getting a dataset and end by serving the trained model in production.

Along the way, you'll see that MLOps is a complex ecosystem, similar to an orchestra: each tool has its own role, but they only make music together.

| Step | What we do | Tool |
|---|---|---|
| 1. Get the data | Download the BUSI ultrasound dataset | DagsHub Data Engine |
| 2. Prepare the data | Explore, clean, deduplicate, split | pandas, scikit-learn |
| 3. Version the data | Track raw and processed data, push them to a remote, record their lineage | DVC + DagsHub |
| 4. Train and track | Train scikit-learn, XGBoost and PyTorch models, log params, metrics, models and lineage | MLflow + DagsHub |
| 5. Compare | Select the best run | MLflow |
| 6. Serve | Package the model and expose it as an API | BentoML |

Code lives on **GitHub**, while **DagsHub** hosts the data remote and the MLflow tracking server, so that every model can be traced back to the exact code and data that produced it.

## Cleaning & formatting data

````python
import re

OUT_DIR = local_dir / "processed"
CLASSES = ["benign", "malignant", "normal"]
SEED = 42
VAL_SIZE, TEST_SIZE = 0.15, 0.15

MASK_RE = re.compile(r"_mask(_\d+)?$", re.IGNORECASE)   # "_mask", "_mask_1", ...
````

````python
import pandas as pd

# Inventory and exclusion
rows = []
for label in CLASSES:
    for p in sorted((RAW_DIR / label).glob("*.png")):
        rows.append({
            "path": p.as_posix(),
            "label": label,
            "is_mask": bool(MASK_RE.search(p.stem)),
        })

all_files = pd.DataFrame(rows)
print(all_files.groupby(["label", "is_mask"]).size().unstack(fill_value=0))

df = all_files[~all_files["is_mask"]].drop(columns="is_mask").reset_index(drop=True)
print(f"\n{len(df)} images kept out of {len(all_files)} files")
````

````python
import hashlib
from PIL import Image

# Duplicates
def file_md5(path, chunk=1 << 20):
    h = hashlib.md5()
    with open(path, "rb") as f:
        for block in iter(lambda: f.read(chunk), b""):
            h.update(block)
    return h.hexdigest()

def img_meta(path):
    with Image.open(path) as im:
        return im.width, im.height, im.mode

df["md5"] = df["path"].map(file_md5)
df[["width", "height", "mode"]] = df["path"].map(img_meta).tolist()
display(df.head())

dup_groups = df[df.duplicated("md5", keep=False)].sort_values("md5")
print(f"{dup_groups['md5'].nunique()} duplicate groups, {len(dup_groups)} files involved")

# Duplicates with conflicting labels: report them and exclude them
conflicts = dup_groups.groupby("md5")["label"].nunique()
conflict_md5 = conflicts[conflicts > 1].index
print(f"{len(conflict_md5)} duplicates with conflicting labels (excluded)")
display(dup_groups.head(10))

df = df[~df["md5"].isin(conflict_md5)]
df = df.drop_duplicates("md5", keep="first").reset_index(drop=True)
print(f"{len(df)} unique images after deduplication")
````

````python
from sklearn.model_selection import train_test_split

# Stratified split
train_val, test = train_test_split(
    df, test_size=TEST_SIZE, stratify=df["label"], random_state=SEED
)
train, val = train_test_split(
    train_val, test_size=VAL_SIZE / (1 - TEST_SIZE),
    stratify=train_val["label"], random_state=SEED,
)

df["split"] = "train"
df.loc[val.index, "split"] = "val"
df.loc[test.index, "split"] = "test"

summary = pd.crosstab(df["label"], df["split"], margins=True)
display(summary)
display(pd.crosstab(df["label"], df["split"], normalize="columns").round(3))
````

````python
# Export
OUT_DIR.mkdir(parents=True, exist_ok=True)
df["label_id"] = df["label"].map({c: i for i, c in enumerate(CLASSES)})
df = df[["path", "label", "label_id", "split", "width", "height", "mode", "md5"]]

df.to_csv(OUT_DIR / "index.csv", index=False)
dup_groups.to_csv(OUT_DIR / "duplicates_report.csv", index=False)
print(f"Index written: {OUT_DIR / 'index.csv'} ({len(df)} rows)")
````

## Data lineage

**Data lineage** answers one question: *which data, transformed by which code, produced this model?*
Today, being able to answer it is non-negotiable:

- **Reproducibility**: retrain exactly the same model months later.
- **Debugging**: when a model misbehaves in production, find out whether the data, the code, or both have changed.
- **Impact analysis**: a labelling error is found in the dataset? Find every model trained on the corrupted version.
- **Compliance**: a diagnosis model must be auditable down to its training data. Regulations such as the EU AI Act require data governance and record-keeping for high-risk AI systems, which include medical devices.

Git and DVC already hold everything we need: the **commit** identifies the code, and the **`.dvc` files** identify the exact version of each dataset through its hash.
We gather this information into a `lineage.json` file that will follow the model all the way to production:

**`lineage.json`** → **MLflow run** (artifact + tags) → **BentoML model** (metadata) → **Bento** → **`model_info` API**

````python
import json
import subprocess
from datetime import datetime, timezone
from pathlib import Path

import yaml


def sh(cmd):
    return subprocess.check_output(cmd, shell=True, text=True).strip()


def dvc_out(dvc_file):
    """Read the hash, size and number of files of a DVC-tracked folder from its .dvc file."""
    out = yaml.safe_load(Path(dvc_file).read_text())["outs"][0]
    return {"dvc_file": str(Path(dvc_file).relative_to(Path.cwd())),
            "md5": out["md5"], "size_bytes": out["size"], "nfiles": out.get("nfiles")}


# No credentials here: the DVC password lives in .dvc/config.local and git uses a credential helper
lineage = {
    "created_at": datetime.now(timezone.utc).isoformat(timespec="seconds"),
    "code": {
        "git_repository": sh("git remote get-url origin"),
        "git_commit": sh("git rev-parse HEAD"),
        "uncommitted_changes": bool(sh("git status --porcelain --untracked-files=no")),
    },
    "data": {
        "source": f"DagsHub datasource {upstream_owner}/{repo_name} (Dataset_BUSI_with_GT)",
        "dvc_remote": sh("dvc config remote.origin.url"),
        "raw": dvc_out(f"{RAW_DIR}.dvc"),
        "processed": dvc_out(f"{OUT_DIR}.dvc"),
    },
    "preparation": {
        "steps": ["drop mask images", "drop duplicates with conflicting labels",
                  "deduplicate on md5", "stratified train/val/test split"],
        "seed": SEED, "val_size": VAL_SIZE, "test_size": TEST_SIZE,
        "classes": CLASSES,
        "split_sizes": {k: int(v) for k, v in df["split"].value_counts().items()},
        "index_md5": hashlib.md5((OUT_DIR / "index.csv").read_bytes()).hexdigest(),
    },
}

if lineage["code"]["uncommitted_changes"]:
    print("WARNING: uncommitted changes, the git commit does not fully describe the code")

Path("lineage.json").write_text(json.dumps(lineage, indent=2))
print(json.dumps(lineage, indent=2))
````

From this file alone, anyone can rebuild the exact training data: `git checkout <git_commit>`, then `dvc pull`.

> `lineage.json` is a derived file: don't commit it, it describes the commit it would belong to. It is stored in MLflow instead.

# Sklearn & XGBoost

## Data formatting

````python
IMG_SIZE = 224                       # torch (pretrained backbone)
SK_SIZE = 64                         # sklearn: small size, flattened vectors
MEAN, STD = [0.485, 0.456, 0.406], [0.229, 0.224, 0.225]
BATCH_SIZE = 32

splits = {s: df[df["split"] == s].reset_index(drop=True) for s in ["train", "val", "test"]}
{s: len(d) for s, d in splits.items()}
````

````python
import numpy as np

def load_sklearn_array(frame, size=SK_SIZE):
    X = np.empty((len(frame), size * size), dtype=np.float32)
    for i, path in enumerate(frame["path"]):
        with Image.open(path) as im:
            im = im.convert("L").resize((size, size), Image.BILINEAR)
        X[i] = np.asarray(im, dtype=np.float32).ravel() / 255.0
    return X, frame["label_id"].to_numpy()

X_train, y_train = load_sklearn_array(splits["train"])
X_val,   y_val   = load_sklearn_array(splits["val"])
X_test,  y_test  = load_sklearn_array(splits["test"])

print(X_train.shape, X_val.shape, X_test.shape)

# Cache to avoid reloading the images at each run
np.savez_compressed(OUT_DIR / f"sklearn_{SK_SIZE}px.npz",
                    X_train=X_train, y_train=y_train,
                    X_val=X_val, y_val=y_val,
                    X_test=X_test, y_test=y_test)
````

## Train the sklearn model with mlflow tracking

````python
import mlflow
from sklearn.metrics import accuracy_score, f1_score, classification_report

INDEX_MD5 = lineage["preparation"]["index_md5"]

def common_params(model_type):
    return {"model_type": model_type, "seed": SEED, "index_md5": INDEX_MD5,
            "n_train": len(splits["train"]), "n_val": len(splits["val"]),
            "n_test": len(splits["test"])}

def log_lineage():
    """Attach the data lineage to the active run: the full file as an artifact, key ids as searchable tags."""
    mlflow.log_artifact("lineage.json")
    mlflow.set_tags({"git_commit": lineage["code"]["git_commit"],
                     "data_raw_md5": lineage["data"]["raw"]["md5"],
                     "data_processed_md5": lineage["data"]["processed"]["md5"]})

def log_eval(y_true, y_pred, prefix):
    metrics = {f"{prefix}_accuracy": accuracy_score(y_true, y_pred),
               f"{prefix}_f1_macro": f1_score(y_true, y_pred, average="macro")}
    mlflow.log_metrics(metrics)
    return metrics

def log_report(y_true, y_pred):
    report = classification_report(y_true, y_pred, target_names=CLASSES)
    mlflow.log_text(report, "classification_report_test.txt")
    print(report)
````

````python
from sklearn.linear_model import LogisticRegression
import time

mlflow.sklearn.autolog(log_input_examples=True, log_datasets=False,
                       log_post_training_metrics=False,
                       serialization_format="skops")

with mlflow.start_run(run_name="sklearn_logreg"):
    mlflow.log_params({**common_params("sklearn_logreg"), "img_size": SK_SIZE})
    log_lineage()

    t0 = time.time()
    clf = LogisticRegression(max_iter=1000, class_weight="balanced", random_state=SEED)
    clf.fit(X_train, y_train)
    mlflow.log_metric("train_time_s", time.time() - t0)

    log_eval(y_val, clf.predict(X_val), "val")
    y_pred = clf.predict(X_test)
    print(log_eval(y_test, y_pred, "test"))
    log_report(y_test, y_pred)

mlflow.sklearn.autolog(disable=True)
````

## Train the XGBoost model

````python
from xgboost import XGBClassifier
from sklearn.utils.class_weight import compute_sample_weight
import time

mlflow.xgboost.autolog(log_input_examples=True, log_datasets=False)

with mlflow.start_run(run_name="xgboost"):
    mlflow.log_params({**common_params("xgboost"), "img_size": SK_SIZE})
    log_lineage()

    t0 = time.time()
    xgb = XGBClassifier(n_estimators=200, max_depth=6, learning_rate=0.1,
                        tree_method="hist", random_state=SEED, n_jobs=-1)
    xgb.fit(X_train, y_train,
            sample_weight=compute_sample_weight("balanced", y_train),
            eval_set=[(X_val, y_val)], verbose=False)   # params, mlogloss curve per iteration and model are logged
    mlflow.log_metric("train_time_s", time.time() - t0)

    log_eval(y_val, xgb.predict(X_val), "val")
    y_pred = xgb.predict(X_test)
    print(log_eval(y_test, y_pred, "test"))
    log_report(y_test, y_pred)

mlflow.xgboost.autolog(disable=True)
````

# PyTorch

## Data formatting

````python
import torch
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms as T

train_tf = T.Compose([
    T.Resize((IMG_SIZE, IMG_SIZE)),
    T.RandomHorizontalFlip(),
    T.RandomRotation(10),
    T.ColorJitter(brightness=0.2, contrast=0.2),
    T.ToTensor(),
    T.Normalize(MEAN, STD),
])

eval_tf = T.Compose([                # = the transform that will go into BentoML
    T.Resize((IMG_SIZE, IMG_SIZE)),
    T.ToTensor(),
    T.Normalize(MEAN, STD),
])

class BusiDataset(Dataset):
    def __init__(self, frame, transform):
        self.paths = frame["path"].tolist()
        self.labels = frame["label_id"].tolist()
        self.transform = transform

    def __len__(self):
        return len(self.paths)

    def __getitem__(self, idx):
        with Image.open(self.paths[idx]) as im:
            img = im.convert("RGB")
        return self.transform(img), self.labels[idx]

datasets_torch = {
    "train": BusiDataset(splits["train"], train_tf),
    "val":   BusiDataset(splits["val"], eval_tf),
    "test":  BusiDataset(splits["test"], eval_tf),
}
````

## Create the loader

````python
g = torch.Generator().manual_seed(SEED)

loaders = {
    s: DataLoader(ds, batch_size=BATCH_SIZE, shuffle=(s == "train"),
                  num_workers=2, generator=g if s == "train" else None)
    for s, ds in datasets_torch.items()
}

# Imbalance benign >> malignant > normal: weights for CrossEntropyLoss
counts = splits["train"]["label_id"].value_counts().sort_index().to_numpy()
class_weights = torch.tensor(counts.sum() / (len(CLASSES) * counts), dtype=torch.float32)
print(dict(zip(CLASSES, class_weights.round(decimals=3).tolist())))

# Check one batch
xb, yb = next(iter(loaders["train"]))
print(xb.shape, xb.dtype, yb[:10])
````

## Load pretrained ResNet-18 and finetune it with early stopping

````python
import copy, os, time
import torch.nn as nn
from torchvision.models import resnet18, ResNet18_Weights

device = "cuda" if torch.cuda.is_available() else "cpu"
MAX_EPOCHS, MIN_EPOCHS, PATIENCE, MIN_DELTA, LR = 50, 15, 8, 1e-3, 1e-3
os.environ["MLFLOW_DEFAULT_PREDICTION_DEVICE"] = "cpu"   # MLflow validates the logged model on CPU


def run_eval(model, loader, criterion):
    """Return (mean loss, y_true, y_pred) over a DataLoader."""
    model.eval()
    total_loss, labels, preds = 0.0, [], []
    with torch.no_grad():
        for xb, yb in loader:
            xb, yb = xb.to(device), yb.to(device)
            logits = model(xb)
            total_loss += criterion(logits, yb).item() * len(xb)
            labels.append(yb.cpu())
            preds.append(logits.argmax(1).cpu())
    return (total_loss / len(loader.dataset),
            torch.cat(labels).numpy(), torch.cat(preds).numpy())


with mlflow.start_run(run_name="torch_resnet18_es"):
    mlflow.log_params({**common_params("torch_resnet18"),
                       "max_epochs": MAX_EPOCHS, "min_epochs": MIN_EPOCHS,
                       "patience": PATIENCE, "min_delta": MIN_DELTA,
                       "early_stopping_metric": "val_loss",
                       "lr": LR, "batch_size": BATCH_SIZE, "img_size": IMG_SIZE,
                       "mean": MEAN, "std": STD, "frozen_backbone": True})
    log_lineage()

    # --- Model: pretrained ResNet-18, frozen backbone, new classification head ---
    torch.manual_seed(SEED)
    model = resnet18(weights=ResNet18_Weights.DEFAULT)
    for p in model.parameters():
        p.requires_grad = False
    model.fc = nn.Linear(model.fc.in_features, len(CLASSES))
    model = model.to(device)

    criterion = nn.CrossEntropyLoss(weight=class_weights.to(device))
    optimizer = torch.optim.Adam(model.fc.parameters(), lr=LR)

    best_loss, best_epoch, best_state = float("inf"), 0, None
    epochs_without_improvement = 0
    t0 = time.time()

    for epoch in range(1, MAX_EPOCHS + 1):
        # --- Training ---
        model.train()
        running = 0.0
        for xb, yb in loaders["train"]:
            xb, yb = xb.to(device), yb.to(device)
            optimizer.zero_grad()
            loss = criterion(model(xb), yb)
            loss.backward()
            optimizer.step()
            running += loss.item() * len(xb)
        train_loss = running / len(loaders["train"].dataset)

        # --- Validation ---
        val_loss, y_true, y_hat = run_eval(model, loaders["val"], criterion)
        val_f1 = f1_score(y_true, y_hat, average="macro")
        mlflow.log_metrics({"train_loss": train_loss, "val_loss": val_loss,
                            "val_f1_macro_epoch": val_f1}, step=epoch)

        # --- Early stopping on validation loss ---
        if val_loss < best_loss - MIN_DELTA:
            best_loss, best_epoch = val_loss, epoch
            best_state = copy.deepcopy(model.state_dict())
            epochs_without_improvement = 0
            flag = "*"
        else:
            epochs_without_improvement += 1
            flag = ""

        print(f"epoch {epoch:02d} | train_loss {train_loss:.3f} | "
              f"val_loss {val_loss:.3f} | val_f1 {val_f1:.3f} {flag}")

        if epoch >= MIN_EPOCHS and epochs_without_improvement >= PATIENCE:
            print(f"Early stopping at epoch {epoch} (best epoch: {best_epoch})")
            break

    mlflow.log_metrics({"train_time_s": time.time() - t0,
                        "best_epoch": best_epoch, "stopped_epoch": epoch,
                        "best_val_loss": best_loss})

    # --- Restore best weights, then final evaluation ---
    model.load_state_dict(best_state)
    _, y_true, y_pred = run_eval(model, loaders["val"], criterion)
    log_eval(y_true, y_pred, "val")
    _, y_true, y_pred = run_eval(model, loaders["test"], criterion)
    print(log_eval(y_true, y_pred, "test"))
    log_report(y_true, y_pred)

    # --- Export as pt2 (traced graph, no pickle) ---
    model.cpu().eval()                                     # eval mode and CPU baked into the graph
    example = next(iter(loaders["test"]))[0][:4].numpy()   # batch > 1 keeps the batch dim dynamic
    mlflow.pytorch.log_model(
        model, name="model", serialization_format="pt2", input_example=example,
        pip_requirements=[f"torch=={torch.__version__.split('+')[0]}"],
    )
````

# BentoML

Our best model is now logged in MLflow, but a logged model is not a product yet: nobody can send it an ultrasound image and get a diagnosis back. This last step turns it into a **prediction service**.

[BentoML](https://docs.bentoml.com) is a framework to package and serve ML models. It fills the gap between the experiment tracker and the production infrastructure:

- **Model store**: a local registry where BentoML keeps the models it can serve.
- **Service**: a Python class that loads the model once and exposes one or more **APIs** (HTTP endpoints). This is where the pre-processing lives: the image must go through exactly the same transform as during evaluation (`eval_tf`), otherwise the predictions are meaningless.
- **Bento**: the deployable artifact. It bundles the service code, the model and its dependencies in a versioned, self-contained archive.
- **Container**: a Bento can be turned into a Docker image with a single command, and then deployed anywhere: a server, Kubernetes, or a cloud platform.

In this section, we will:

1. **Select** the best PyTorch run from MLflow.
2. **Import** its model from the DagsHub tracking server into the BentoML model store, together with its **data lineage**.
3. **Serve** it locally and send it a test request.
4. **Build** a Bento and **containerize** it, ready for production.

> MLflow credentials are only needed in step 2: the model and its lineage are downloaded once and then copied into the Bento. The deployed service is self-contained and never connects to MLflow.

## Select the best model

Our service loads a PyTorch model exported as `pt2`, so we only consider the `torch*` runs and keep the best one on the test macro F1.

The service pre-processing is defined in `bentoml/preprocess.py`. We check that it matches the parameters logged with the run: a mismatch would silently degrade the predictions in production (*training/serving skew*).

````python
import sys
sys.path.insert(0, "bentoml")            # make bentoml/preprocess.py importable
import preprocess as serving_pp

best = mlflow.search_runs(
    experiment_names=["busi-classification"],
    filter_string="tags.mlflow.runName LIKE 'torch%' AND attributes.status = 'FINISHED'",
    order_by=["metrics.test_f1_macro DESC"],
    max_results=1,
).iloc[0]
run_id = best["run_id"]
model_uri = f"runs:/{run_id}/model"
print(f"Best run: {best['tags.mlflow.runName']} ({run_id}) - test_f1_macro={best['metrics.test_f1_macro']:.4f}")

# Training/serving skew check
params = mlflow.get_run(run_id).data.params
expected = {"img_size": str(serving_pp.IMG_SIZE), "mean": str(serving_pp.MEAN), "std": str(serving_pp.STD)}
for key, value in expected.items():
    assert params.get(key) in (None, value), f"{key}: run={params.get(key)} vs preprocess.py={value}"
assert serving_pp.CLASSES == CLASSES
print("bentoml/preprocess.py matches the training pre-processing")
````

## Import the model into the BentoML model store

`bentoml.mlflow.import_model` downloads the model from the MLflow server (using the environment variables set in the setup section) and stores it locally under a versioned tag.
We attach metadata pointing back to the MLflow run, and the **lineage of the run**, read from its `lineage.json` artifact.
The metadata is stored with the model, so it ends up in the Bento: any deployed model can tell which experiment, code and data produced it.

````python
import bentoml
from pathlib import Path

# Lineage of the selected run (not the current one: the best run may come from an older data version)
run_lineage = mlflow.artifacts.load_dict(f"runs:/{run_id}/lineage.json")
print("Trained on commit", run_lineage["code"]["git_commit"],
      "with processed data", run_lineage["data"]["processed"]["md5"])

bento_model = bentoml.mlflow.import_model(
    "busi_resnet18",
    model_uri=model_uri,
    labels={"dataset": "BUSI", "framework": "pytorch", "format": "pt2"},
    metadata={
        "mlflow_tracking_uri": os.environ["MLFLOW_TRACKING_URI"],
        "mlflow_model_uri": model_uri,
        "mlflow_run_id": run_id,
        "test_f1_macro": float(best["metrics.test_f1_macro"]),
        "classes": ",".join(CLASSES),
        "img_size": IMG_SIZE,
        "lineage": run_lineage,
    },
)

# Layout of an MLflow PyTorch model saved with serialization_format="pt2"
assert Path(bento_model.path_of("mlflow_model/data/model.pt2")).exists(), \
    f"{model_uri} is not a pt2 model: log it with serialization_format='pt2'"
print("Imported into the BentoML store:", bento_model.tag)
````

## Serve locally

`bentoml serve` starts an HTTP server exposing the APIs of the service. We launch it in the background so that the notebook stays usable.
If you're running locally (not on Colab), you can also open http://localhost:3000 to try the APIs from the Swagger UI.

````python
import subprocess

serve_log = open("/tmp/bentoml_serve.log", "w")
server = subprocess.Popen(
    ["bentoml", "serve", "service:BusiClassifier", "--port", "3000"],
    cwd="bentoml", stdout=serve_log, stderr=subprocess.STDOUT,
)
````

We send an image from the test set to the service, exactly as a client application would. If the request fails, check the server logs with `!tail -n 50 /tmp/bentoml_serve.log`.

````python
sample = splits["test"].sample(1, random_state=SEED).iloc[0]

with bentoml.SyncHTTPClient("http://localhost:3000", server_ready_timeout=120) as client:
    print("Model info:", json.dumps(client.model_info(), indent=2))
    print("True label:", sample["label"])
    print("Prediction:", client.predict(image=Path(sample["path"])))
````

The `model_info` API returns the lineage of the served model: from a running service, we can go back to the git commit and the DVC data versions that produced it.

````python
# Stop the server
server.terminate()
server.wait()
serve_log.close()
````

## (Optional) Serve the built Bento

Earlier, we served the service **from its source code** (development mode). We can also serve the **Bento** we just built: this tests the exact artifact that will be deployed, and catches packaging mistakes such as a file missing from the `include` list of `bentofile.yaml`.

> The Bento still runs in the Python environment of this notebook: its dependencies (`requirements.txt`) are only truly tested once containerized.

````python
serve_log = open("/tmp/bentoml_serve.log", "w")
server = subprocess.Popen(
    ["bentoml", "serve", "busi_classifier:latest", "--port", "3000"],
    stdout=serve_log, stderr=subprocess.STDOUT,
)

try:
    with bentoml.SyncHTTPClient("http://localhost:3000", server_ready_timeout=120) as client:
        print("True label:", sample["label"])
        print("Prediction:", client.predict(image=Path(sample["path"])))
finally:                                 # always stop the server, even if the request fails
    server.terminate()
    server.wait()
    serve_log.close()
````
